```python
import openpyxl
import time

wb = openpyxl.Workbook()
ws = wb.active

title = ws.cell(row=2, column=2, value='請求書')

info = [
       '株式会社ABC',
       '〒101-0022 東京都千代田区神田練塀町300',
       'TEL:03-1234-5678 FAX:03-1234-5678',
       '担当者名:鈴木一郎 様']

for i, text in enumerate(info):
    ws.cell(
        row=title.row + 2 + i,
        column=2,
        value=text
    )


product_headers = ['商品名', '数量', '単価', '金額']

for i, text in enumerate(product_headers):
    ws.cell(
        row=title.row + 8,
        column=title.column + i,
        value=text
    )


data = [
    [None, '商品A', 2, 10000, 20000],
    [None, '商品B', 1, 15000, 15000],
    [None, None, None, None, 35000],
    [None],
    [None, '合計', None, None, 35000],
    [None, '消費税', None, None, 3500],
    [None,'税込み合計', None, None, 38500]
]

for row in data:
    ws.append(row)


records = [
    ['No.', '0001'],
    ['日付',time.strftime('%Y/%m/%d')]
]

for r, row in enumerate(records):
    for c, value in enumerate(row):
        ws.cell(
            row=title.row + 2 + r,
            column=title.column + 4 + c,
            value=value
        )


wb.save('請求書_20260917.xlsx')
```
